# Objetivo

Realizar uma exploração inicial dos dados para compreender como as tabelas estão estruturadas e identificar:

* Tipos de dados;
* Dimensões (linhas e colunas) das tabelas;
* Nomes e estruturas das colunas;
* Relacionamentos entre as tabelas;
* Colunas relevantes para a análise;
* Valores nulos;
* Valores duplicados.

# Importações

```python
import pandas as pd

from pipeline_vendas.extract import extract_raw_data
```

# Carregamento dos Dados

```python
dfs = extract_raw_data()
```

# Exploração dos Dados

```python
# Mostra o nome e o shape de cada DataFrame
for name, df in dfs.items():
    print(f"{name}: {df.shape}")
```

```python
# Mostra as cinco primerias linhas de cada DataFrame
for name, df in dfs.items():
    print(f"\n===== {name.upper()} =====")
    display(df.head(5))
```

```python
# Mostra as informações de cada DataFrame
for name, df in dfs.items():
    print(f"\n===== {name.upper()} =====")
    df.info()
```

```python
# Mostra a quantidade de valores nulos e duplicados de cada DataFrame
for name, df in dfs.items():
    print(f"\n===== {name.upper()} =====")
    print(f"Quantidade Nulos:\n{df.isna().sum()[df.isna().sum() > 0]}")
    print(f"\nQuantidade de Duplicados: {df.duplicated().sum()}")
```

# Conclusão

* Não foram identificadas linhas completamente duplicadas nas tabelas;
* As tabelas `Orders` e `Products` possuem valores ausentes.

Os relacionamentos entre as tabelas foram definidos com base na documentação disponibilizada junto ao dataset, considerando as chaves e a estrutura dos dados.

Foram identificados os seguintes relacionamentos:

* `Customers (1) ----> Orders (0, N)`
* `Products (1) ----> Order Items (1, N)`
* `Sellers (1) ----> Order Items (1, N)`
* `Orders (1) ----> Order Items (1, N)`
