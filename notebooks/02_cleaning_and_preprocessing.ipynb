# Objetivo

Realizar a limpeza e o tratamento dos dados, preparando-os para as etapas posteriores do projeto.

* Tratar os valores nulos;
* Remover colunas que não fazem parte do escopo do projeto;
* Exportar os dados processados para a próxima etapa da pipeline.

# Importações

```python
import pandas as pd

from pipeline_vendas.extract import extract_raw_data
from pipeline_vendas.cleaning import select_relevant_columns
from pipeline_vendas.load import load_processed_data
```

# Carregamento dos Dados

```python
dfs = extract_raw_data()
```

# Limpeza e Tratamento

```python
# Remove as colunas que não fazem parte do escopo do projeto
dfs = select_relevant_columns(dfs)
```

```python
# Substitui valores nulos da categoria do produto por "outros"
dfs["products"]["product_category_name"] = (
    dfs["products"]["product_category_name"].fillna("outros")
)
```

# Validação

```python
# Verifica se ainda tem algum valor nulo nas tabelas
for name, df in dfs.items():
    print(f"\n===== {name.upper()} =====")
    print(df.isna().sum()[df.isna().sum() > 0])
```

# Exportação dos Dados

```python
# Carrega os dados limpos para a pasta processed
load_processed_data(dfs)
```

# Conclusão

Durante a etapa de limpeza e seleção das variáveis, foram removidas as colunas que não fazem parte do escopo definido para o projeto.

### Customers

Foi removida a seguinte coluna:

* `customer_zip_code_prefix`

### Sellers

Foi removida a seguinte coluna:

* `seller_zip_code_prefix`

### Products

Foram mantidas apenas as colunas:

* `product_id`
* `product_category_name`

Consequentemente, foram removidas as seguintes colunas:

* `product_name_lenght`
* `product_description_lenght`
* `product_photos_qty`
* `product_weight_g`
* `product_length_cm`
* `product_height_cm`
* `product_width_cm`

### Orders

Foram removidas as seguintes colunas:

* `order_approved_at`
* `order_delivered_carrier_date`
* `order_delivered_customer_date`
* `order_estimated_delivery_date`

Os valores nulos identificados em `Orders` estavam presentes nas colunas relacionadas às etapas de aprovação e entrega dos pedidos. Como essas variáveis não fazem parte do escopo definido para o projeto, as respectivas colunas foram removidas.

### Order Items

Foi removida a seguinte coluna:

* `shipping_limit_date`

Além da seleção das variáveis, os valores nulos presentes em `product_category_name` foram substituídos por **`'outros'`**, preservando os registros dos produtos e representando aqueles cuja categoria não foi informada no conjunto de dados.

Após o tratamento, foi realizada uma verificação dos valores ausentes para validar o resultado da etapa de limpeza.

Os dados resultantes foram exportados para a pasta `processed`, onde serão utilizados como entrada para a etapa de transformação e modelagem.
